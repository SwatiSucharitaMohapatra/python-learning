# CLASSIFICATION

A classifier is a model that predicts to which class a sample belongs, based on the sample features.

For example, the logistic regression you saw in the Statistics course is a classifier. It estimates the probability that a sample belongs to the positive class (in a 2-class problem), using the sigmoid function.

* Create a function `sigmoid`
  * Argument: `x` a float number
  * Returns: `x` a float number
  * **bonus**: use type hinting

The sigmoid function is: $\sigma(x) = \frac{1}{1 + e^{-x}}$

```python
print(math.isclose(sigmoid(0.0), 0.5))

```

````python
import math
def sigmoid (x: float) -> float :
    x = 1 / (1 + math.exp(-x))
    return x

````

````python
print(math.isclose(sigmoid(0.0), 0.5))
print(math.isclose(sigmoid(2.0), 0.8807970779778823))
print(math.isclose(sigmoid(-1.0), 0.2689414213699951))
````

## Exercise 3 - Scoring

* A classifier works like this:
  * Based on the features of a sample (1 feature = 1 number)
  * Based on coefficients (1 coefficient = 1 number)
  * It computes the sum of the products of each coefficient by the feature at the same index
  * e.g. `coefs = [3, 9], features = [1, 7], raw_score = 3*1 + 9*7 = 66`
  * Then it applies the sigmoid function to get a number between 0.0 and 1.0
  * When the number is higher than `0.5` it predicts that the sample belongs to class `1`, otherwise it is `0`
* Create a function `score`
  * Argument 1: `coefs` a list of float numbers
  * Argument 2: `features` a list of float numbers
  * Returns a float number
  * Returns the sigmoid function applied to the raw score
  * **bonus:** use type hinting

You have it right when this displays `True`
```python
print(math.isclose(score([ 1.8, 0.1, -0.5, 0.4], X[0]), 0.9791150816064992))
```

````python
# score = dot-product of 2 lists of numbers of the same length
import math
def score (coefs: list, features: list) -> float :
    raw_score = 0
    for i in range(len(coefs)):
        raw_score += coefs[i] * features[i]
    ran_num = 1 / (1 + math.exp(-raw_score))
    return ran_num

````

````python
print(math.isclose(score([ 1.8, 0.1, -0.5, 0.4], X[0]), 0.9791150816064992))
````

## Exercise 4 - Classifying

* Finish the classifier
* Create a function `classify`
  * Argument 1: `coefs` list of float numbers
  * Argument 2: `features` list of float numbers
  * Returns an integer, either `0` or `1`
  * The return value is the predicted class for the sample's feature
  * **bonus:** use type hinting

You have it right when this display `True`

```python
print(classify([ 1.8, 0.1, -0.5, 0.4], X[0]) == 1)
print(classify([ 1.8, 0.1, -0.5, 0.4], X[2]) == 0)
```

````python
def classify (coefs: list, features: list) -> int :
    score_output = score(coefs, features)
    return 1 if score_output > 0.5 else 0
````

````python
print(classify([ 1.9, 0.1, 0.5, 1.0], X[0]) == 1)
print(classify([ 1.8, 0.1, -0.5, 0.4], X[2]) == 0)
````

## Exercise 5 - Predict

* Create a function `predict`
  * Argument 1: `coefs`: list of float numbers
  * Argument 2: `matrix`: a numpy array (i.e. a matrix with N rows, and D columns)
  * `matrix` array has N rows (1 row = 1 sample)
  * `matrix` array has D columns (1 column = 1 feature)
  * Returns: a list of integers
  * Returns the classification for each sample

You have it right when the following code displays `True`

```python
print(predict([ 1.8, 0.1, -0.5, 0.4], [X[0], X[2]]) == [1, 0])
```

````python
import numpy as np

def predict(coefs: list, matrix: np.ndarray) -> list:
    predictions = []
    for sample_features in matrix:
        predictions.append(classify(coefs, sample_features.tolist()))
    return predictions
````

````python
print(predict([ 1.8, 0.1, -0.5, 0.4], [X[0], X[2]]) == [1, 0])
````
